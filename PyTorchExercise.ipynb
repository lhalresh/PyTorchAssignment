Part 9 — Automatic Differentiation

PyTorch provides automatic differentiation using autograd.

````python
x = torch.tensor([[1., 2.],
                  [3., 4.]], requires_grad=True)

# TODO: Calculate y = 3x² + 2
y = 3 * x ** 2 + 2
print("y = 3x² + 2:")
print(y)
````

````python
# Then calculate:
z = torch.sum(y)

# Use:
z.backward()
# to calculate: dz/dx


# Print:
print("y:")
print(y)
print("z:")
print(z)
print("x.grad:")
print(x.grad)
````

Questions
* What does requires_grad=True mean?
* What does backward() do?
* What is stored in x.grad?
* Why is automatic differentiation important for neural networks?

### tells PyTorch to track the operations performed on a tensor so that PyTorch can calculate its gradients later
### calculates the gradients of the output with respect to the tensors that have requires_grad=True
### x.grad stores the derivative of z with respect to each element of x
### Automatic differentiation allows PyTorch to efficiently calculate these gradients,
