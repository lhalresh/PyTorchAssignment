## Part 3 — Boolean Indexing

Create:
```python
x = np.array([[12, 5, 18, 3],
              [25, 7, 14, 9],
              [2, 30, 11, 20]])
```
1.	Extract all values greater than 15.
2.	Extract all values less than or equal to 10.
3.	Replace every value greater than 20 with 0.
4.	Count how many values are greater than 15.

````python
X = np.array([
    [12, 5, 18, 31],
    [25, 7, 14, 9],
    [2, 30, 11, 20]
])

# 1. 
print("1. Values greater than 15:")
print(X[X > 15])
print()

# 2. 
print("2. Values less than or equal to 10:")
print(X[X <= 10])
print()

# 3. 
X_copy = X.copy()
X_copy[X_copy > 20] = 0

print("3. Values greater than 20 replaced with 0:")
print(X_copy)
print()

# 4.
count = np.sum(X > 15)

print("4. Number of values greater than 15:")
print(count)
````

## Part 5 — Matrix Multiplication
Create:
```python
A = np.array([[1, 2, 3],
              [4, 5, 6]])

B = np.array([[7, 8],
              [9, 10],
              [11, 12]])
```
1.	Compute A @ B.
2.	Compute the same result using np.dot(A, B).
3.	Verify that both results are identical.
4.	Print the shape of A, B, and the result.

The notebook demonstrates the equivalence of .dot(), np.dot(), and @ for matrix multiplication.

````python
A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

B = np.array([
    [7, 8],
    [9, 10],
    [11, 12]
])

# 1. 
result1 = A @ B

print("1. A @ B:")
print(result1)
print()

# 2.
result2 = np.dot(A, B)

print("2. np.dot(A, B):")
print(result2)
print()

# 3. 
print("3. Are the results identical?")
print(np.array_equal(result1, result2))
print()

# 4. 
print("4. Shapes:")
print("Shape of A:", A.shape)
print("Shape of B:", B.shape)
print("Shape of result:", result1.shape)
````

## Part 7 — Mini Application: Student Grades

Create a NumPy array containing the grades of 5 students in 4 assignments:

* Student 1: 85, 90, 78, 92
* Student 2: 70, 88, 91, 76
* Student 3: 95, 93, 89, 97
* Student 4: 60, 72, 68, 75
* Student 5: 88, 84, 90, 86

Using NumPy only:
1.	Calculate each student’s average.
2.	Find the highest grade on each assignment.
3.	Find the average grade for each assignment.
4.	Find the student with the highest overall average.
5.	Find all grades below 70.
6.	Replace grades below 70 with 70.
7.	Calculate the overall class average.

````python
grades = np.array([
    [85, 90, 78, 92],  # Student 1
    [70, 88, 91, 76],  # Student 2
    [95, 93, 89, 97],  # Student 3
    [60, 72, 68, 75],  # Student 4
    [88, 84, 90, 86]   # Student 5
])

print("Grades:")
print(grades)
print()

# 1. 
student_averages = np.mean(grades, axis=1)

print("1. Student averages:")
print(student_averages)
print()

# 2. 
highest_grades = np.max(grades, axis=0)

print("2. Highest grade on each assignment:")
print(highest_grades)
print()

# 3. 
assignment_averages = np.mean(grades, axis=0)

print("3. Average grade for each assignment:")
print(assignment_averages)
print()

# 4. 
highest_student = np.argmax(student_averages) + 1

print("4. Student with the highest overall average:")
print("Student", highest_student)
print()

# 5.
below_70 = grades[grades < 70]

print("5. Grades below 70:")
print(below_70)
print()

# 6. 
updated_grades = grades.copy()
updated_grades[updated_grades < 70] = 70

print("6. Grades after replacing values below 70:")
print(updated_grades)
print()

# 7.
class_average = np.mean(updated_grades)

print("7. Overall class average:")
print(class_average)
print()
````

## Challenge Problem

Create a random 5 × 5 matrix containing integers from 1 to 100.

Using NumPy only:
1.	Find the largest value.
2.	Find the smallest value.
3.	Calculate the mean.
4.	Calculate the standard deviation.
5.	Find all values greater than the mean.
6.	Replace all values below the mean with 0.
7.	Calculate the sum of each row and each column.

````python
np.random.seed(42)

challenge = np.random.randint(1, 101, size=(5, 5))

print("Random 5 x 5 matrix:")
print(challenge)
print()

# 1. 
largest = np.max(challenge)
print("1. Largest value:", largest)
print()

# 2. 
smallest = np.min(challenge)
print("2. Smallest value:", smallest)
print()

# 3. 
mean_value = np.mean(challenge)
print("3. Mean:", mean_value)
print()

# 4. 
std_value = np.std(challenge)
print("4. Standard deviation:", std_value)
print()

# 5. 
greater_than_mean = challenge[challenge > mean_value]
print("5. Values greater than the mean:")
print(greater_than_mean)
print()

# 6. 
modified = challenge.copy()
modified[modified < mean_value] = 0

print("6. Matrix after replacing values below the mean:")
print(modified)
print()

# 7. 
row_sums = np.sum(challenge, axis=1)
column_sums = np.sum(challenge, axis=0)

print("7. Sum of each row:")
print(row_sums)

print("Sum of each column:")
print(column_sums)
````
